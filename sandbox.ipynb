```python
bm = np.array([0, 0, 1.413, 1.413, 7.577, 1.136, 8.289, 53.904, 73.629])
bm = bm / np.max(bm)

t= np.array([0,0, 1,1, 4, 4, 9, 14, 15])
t = t / np.max(t)

# Fit exponential function: bm = a * exp(b * t)
from scipy.optimize import curve_fit

# Define exponential function
def exponential(t, a, b):
    return a * np.exp(b * t)

# Filter out zero bm values for better fit
mask = bm > 0
t_fit = t[mask]
bm_fit = bm[mask]

# Fit the curve
params, covariance = curve_fit(exponential, t_fit, bm_fit, p0=[1, 0.1])
a_fit, b_fit = params

print(f"Fitted exponential: bm = {a_fit:.4f} * exp({b_fit:.4f} * t)")

# Generate smooth curve for plotting
t_smooth = np.linspace(t.min(), t.max(), 100)
bm_smooth = exponential(t_smooth, a_fit, b_fit)
bm_smooth = exponential(t_smooth, 0.0078, 4.85)
fig, ax = plt.subplots()
ax.plot(t, bm, 'o', label='Data')
ax.plot(t_smooth, bm_smooth, '-', label=f'Fit: $bm = {a_fit:.2f} e^{{{b_fit:.3f}t}}$')
ax.set_xlabel('t')
ax.set_ylabel('bm')
ax.legend()
ax.grid()
plt.show()
```
